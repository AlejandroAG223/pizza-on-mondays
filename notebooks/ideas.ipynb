```python
import yfinance as yf
```

```python
stocks = ["ADC", "AKR", "BRX", "EPRT", "FCPT", "KIM", "KRG", "MAC", "NNN", "O", "PECO", "REG", "UE"] # "CURB"
```

```python
data = yf.download(stocks, start="2020-01-01", end="2026-12-31")["Close"]
```

```python
data = data.ffill() #NO miedo
```

```python
data = data.bfill() #miedo te crea un lookahead bias
```
