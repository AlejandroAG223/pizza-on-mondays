```python
import yfinance as yf
import seaborn as sns
```

```python
stocks = ["ADC", "AKR", "BRX", "EPRT", "FCPT", "KIM", "KRG", "MAC", "NNN", "O", "PECO", "REG", "UE"] # "CURB"
```

```python
data_rstate = yf.download(stocks, start="2026-01-01", end="2026-12-31")["Close"]
data_rstate = data_rstate.ffill() #NO miedo
data_rstate = data_rstate.bfill() #miedo te crea un lookahead bias
retornos_state = data_rstate.pct_change().dropna()
```

```python
retornos_state.describe()
```

```python
sns.pairplot(retornos_state, kind="scatter", plot_kws={"alpha": 0.5})
```
